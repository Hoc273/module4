# Quick Summary Numbers (prep for Day 2 visualizations)
Loads the already-cleaned dataset (`AB_NYC_2019-clean.csv`) and computes the numbers we'll want on hand before building charts: distribution of listings by group/type, core summary stats, and price by borough.

## 1. Load the cleaned data

```python
import pandas as pd

df = pd.read_csv("AB_NYC_2019-clean.csv")
print(f"Rows: {df.shape[0]}, Columns: {df.shape[1]}")
```

## 2. Listings by `neighbourhood_group` and `room_type`
Count and percentage of total listings for each category, so we know relative sizes before charting (e.g. bar widths, whether a pie chart is even reasonable).

```python
def distribution(column):
    counts = df[column].value_counts()
    pct = (counts / len(df) * 100).round(2)
    result = pd.DataFrame({"count": counts, "percent": pct})
    result.index.name = column
    return result

print("Listings by neighbourhood_group:")
display(distribution("neighbourhood_group"))

print("\nListings by room_type:")
display(distribution("room_type"))
```

## 3. Summary stats for `price`, `minimum_nights`, `number_of_reviews`
Min, max, mean, and median for each numeric column of interest.

```python
summary_cols = ["price", "minimum_nights", "number_of_reviews"]

summary_stats = pd.DataFrame({
    "min": df[summary_cols].min(),
    "max": df[summary_cols].max(),
    "mean": df[summary_cols].mean().round(2),
    "median": df[summary_cols].median(),
})
summary_stats.index.name = "column"
display(summary_stats)
```

## 4. Average and median price by `neighbourhood_group`
Grouped price stats, sorted highest average first — useful to spot which boroughs skew expensive before plotting.

```python
price_by_group = df.groupby("neighbourhood_group")["price"].agg(
    average_price="mean", median_price="median"
).round(2).sort_values("average_price", ascending=False)

display(price_by_group)
```
